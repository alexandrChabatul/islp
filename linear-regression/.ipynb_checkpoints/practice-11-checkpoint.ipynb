```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
from ISLP.models import (ModelSpec as MS, summarize,poly)
from matplotlib.pyplot import subplots
from statsmodels.stats.anova import anova_lm
import math
```

```python
rng = np.random.default_rng(1)
x = rng.normal(size=100)
y = 2 * x + rng.normal(size=100)
y
```

```python
x
```

```python
results = sm.OLS(y,x).fit()
summarize(results)
```

Как мы видим и результаток коэффициент В1 для х1 в целом определен достаточно близко к оптимальному углу наклона.

```python
results1 = sm.OLS(x,y).fit()
summarize(results1)
```

Визуально Коээфициенты сильно отличаются, но можно заметить, что Т-значение одинаково

```python
x1 = pd.DataFrame(x)
x1['intercept'] = 1;
results2 = sm.OLS(y,x1).fit()
summarize(results2)
```

```python
y1 = pd.DataFrame(y)
y1['intercept'] = 1;
results3 = sm.OLS(x,y1).fit()
summarize(results3)
```
