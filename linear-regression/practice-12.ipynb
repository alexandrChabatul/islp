```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
from ISLP.models import (ModelSpec as MS, summarize,poly)
from matplotlib.pyplot import subplots
from statsmodels.stats.anova import anova_lm
import math

np.random.seed(42)

n = 100

X = np.random.normal(loc=10, scale=5, size=n)
Y = 2 * X + np.random.normal(loc=0, scale=10, size=n)

model_Y_given_X = sm.OLS(Y, X).fit()
beta_Y_given_X = model_Y_given_X.params[0]

model_X_given_Y = sm.OLS(X, Y).fit()
beta_X_given_Y = model_X_given_Y.params[0]

# Print results
print("Coefficient of regression Y on X (beta_Y_given_X):", beta_Y_given_X)
print("Coefficient of regression X on Y (beta_X_given_Y):", beta_X_given_Y)
```
